```python
import pandas as pd
from sklearn.datasets import load_iris
iris = load_iris()
X = iris.data
y = iris.target
df = pd.DataFrame(X, columns=iris.feature_names)
df['Species'] = iris.target_names[y]
print("Iris Dataset Loaded Successfully!")
print(df.head())
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)
print("Machine Learning Model Trained Successfully!")
```

```python
import joblib
joblib.dump(model, 'iris_model.pkl')
print("Model Saved Successfully!")
```

```python
import gradio as gr
def predict_flower(sl, sw, pl, pw):
    prediction = model.predict([[sl, sw, pl, pw]])
    species = ['Setosa', 'Versicolor', 'Virginica']
    return species[prediction[0]]
app = gr.Interface(
    fn=predict_flower,
    inputs=[
        gr.Slider(4.0, 8.0, value=5.0, label="Sepal Length"),
        gr.Slider(2.0, 4.5, value=3.0, label="Sepal Width"),
        gr.Slider(1.0, 7.0, value=2.0, label="Petal Length"),
        gr.Slider(0.1, 2.5, value=1.0, label="Petal Width")
    ],
    outputs="text",
    title="Iris Flower Prediction App",
    description="Enter flower measurements to predict its species."
)

print("Gradio App Created Successfully!")
```

```python
app.launch(share=True)
```
