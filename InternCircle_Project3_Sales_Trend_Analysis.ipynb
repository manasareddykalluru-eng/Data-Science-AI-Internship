```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print("Libraries imported successfully!")
```

```python
data = {
    'Order_ID': [101,102,103,104,105,106,107,108,109,110],
    'Customer': ['A','B','C','A','D','B','E','C','A','D'],
    'Product': ['Laptop','Phone','Mouse','Laptop','Keyboard',
                'Phone','Monitor','Mouse','Laptop','Keyboard'],
    'Quantity': [1,2,3,1,2,1,2,4,1,3],
    'Price': [50000,20000,500,50000,1500,20000,12000,500,50000,1500],
    'Date': ['2026-01-05','2026-01-10','2026-02-02',
             '2026-02-15','2026-03-01','2026-03-10',
             '2026-04-05','2026-04-15','2026-05-01','2026-05-10']
}
df = pd.DataFrame(data)
df['Date'] = pd.to_datetime(df['Date'])
df['Revenue'] = df['Quantity'] * df['Price']
print("Sales Dataset Created Successfully!")
df.head(10)
```

```python
print("Dataset Shape:")
print(df.shape)
print("\nDataset Information:")
df.info()
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
total_revenue = df['Revenue'].sum()
print("Total Revenue:", total_revenue)
```

```python
product_revenue = df.groupby('Product')['Revenue'].sum()
print(product_revenue)
```

```python
product_revenue.plot(kind='bar', figsize=(8,5))
plt.title('Product-wise Revenue')
plt.xlabel('Product')
plt.ylabel('Revenue')
plt.show()
```

```python
top_customers = df.groupby('Customer')['Revenue'].sum().sort_values(ascending=False)
print("Top Customers:")
print(top_customers)
```

```python
top_customers.plot(kind='bar', figsize=(8,5))
plt.title('Top Customers by Revenue')
plt.xlabel('Customer')
plt.ylabel('Total Revenue')
plt.show()
```

```python
df['Month'] = df['Date'].dt.month_name()
monthly_revenue = df.groupby('Month')['Revenue'].sum()
print(monthly_revenue)
```

```python
top_customers = df.groupby('Customer')['Revenue'].sum().sort_values(ascending=False)
print("Top Customers:")
print(top_customers)
```

```python
top_customers.plot(kind='bar', figsize=(8,5))
plt.title('Top Customers by Revenue')
plt.xlabel('Customer')
plt.ylabel('Total Revenue')
plt.show()
```

```python
df['Month'] = df['Date'].dt.month_name()
monthly_revenue = df.groupby('Month')['Revenue'].sum()
print("Monthly Revenue:")
print(monthly_revenue)
```

```python
monthly_revenue.plot(kind='line', marker='o', figsize=(8,5))
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.grid(True)
plt.show()
```

```python
customer_segments = df.groupby('Customer').agg({
    'Revenue': 'sum',
    'Order_ID': 'count'
})
customer_segments.columns = ['Total_Revenue', 'Total_Orders']
print("Customer Segmentation:")
print(customer_segments)
```

```python
print("===== SALES ANALYSIS SUMMARY =====")
print("Total Revenue:", df['Revenue'].sum())
print("Total Orders:", df['Order_ID'].nunique())
print("Total Customers:", df['Customer'].nunique())
print("Top Customer:", top_customers.idxmax())
print("Best Selling Product:",
      df.groupby('Product')['Quantity'].sum().idxmax())
```
